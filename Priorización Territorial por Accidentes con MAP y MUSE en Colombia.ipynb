### **Priorización territorial por accidentes con MAP/MUSE en Colombia**

### **1. Introducción**

Los accidentes por **Minas Antipersonal (MAP)** y **Municiones sin Explosionar (MUSE)** constituyen una problemática humanitaria persistente en **Colombia**, especialmente en territorios rurales, étnicos y de frontera históricamente afectados por dinámicas de conflicto armado. La documentación oficial del Estado colombiano reconoce que, tras la firma del Acuerdo de Paz de 2016, el uso de minas antipersonal por parte de grupos armados ilegales persiste en regiones donde el conflicto sigue activo, con altos niveles de afectación sobre la población civil y especiales dificultades de acceso para las labores de intervención (Gobierno de Colombia, 2025a, 2025b). 

Estos eventos generan riesgos directos para la vida, la integridad física y la movilidad de la población civil, además de limitar el acceso a medios de vida, servicios básicos, actividades productivas y presencia institucional en zonas contaminadas o con sospecha de contaminación. Distintos reportes han documentado que la presencia de MAP/MUSE restringe el acceso seguro a salud, educación, agua, tierras productivas y actividades generadoras de ingreso, favorece el confinamiento y el desplazamiento, y obstaculiza tanto la acción humanitaria como la llegada oportuna de servicios estatales a los territorios afectados (ACAPS, 2022; Comité Internacional de la Cruz Roja, 2017, 2025a, 2025b; Foro Humanitario Colombia, 2025). 

La afectación por MAP y MUSE no se distribuye de manera homogénea en el territorio nacional; por el contrario, tiende a concentrarse en determinados departamentos y municipios donde confluyen factores como presencia histórica de actores armados, disputa territorial, ruralidad, economías ilícitas, restricciones de acceso y debilidad institucional. El Landmine Monitor reportó para 2023 contaminación por minas antipersonal en 71 municipios de 14 departamentos, mientras que análisis recientes sobre Colombia muestran concentración de víctimas y población en necesidad en departamentos como Antioquia, Cauca, Norte de Santander, Nariño y Chocó, así como en municipios rurales sometidos a alta disputa armada (ACAPS, 2022; Foro Humanitario Colombia, 2025; Landmine and Cluster Munition Monitor, 2024). En esa misma línea, la solicitud oficial de extensión del plazo de desminado advierte que la falta de información precisa sobre la localización de áreas contaminadas y la ausencia de análisis de contexto sobre las dinámicas del conflicto limitan la planeación, la priorización y la intervención; por ello, **analizar la distribución espacial y temporal de los accidentes permite identificar patrones de concentración, persistencia y posible priorización territorial** (Gobierno de Colombia, 2025a, 2025b). 

El análisis cuantitativo y espacial de crisis humanitarias y contextos de conflicto ha cobrado relevancia en la literatura especializada, en la medida en que la analítica geoespacial permite pasar de la cartografía descriptiva a la identificación de clústeres, patrones y modelos predictivos útiles para la coordinación y la toma de decisiones. En el campo humanitario, se ha señalado que el análisis geoespacial puede mejorar el rigor analítico y la eficiencia programática; y, en desminado humanitario, que la combinación de aprendizaje automático y sistemas de información geográfica puede apoyar la predicción de áreas minadas y la priorización de su despeje (Greenough & Nelson, 2019; Saliba et al., 2024). 

En este contexto, **este proyecto busca aplicar técnicas de análisis exploratorio, estimación de densidad espacial, agrupamiento no supervisado y clasificación supervisada para caracterizar los accidentes por MAP y MUSE en Colombia.** La base utilizada contiene **7.474** registros con información sobre tipo de evento, departamento, municipio, año, mes, coordenadas geográficas y tipo de área. Además, cubre eventos entre **1990 y 2026**, con registros distribuidos en **32 departamentos** y **501 municipios**. A partir de esta información, el análisis se orienta a responder tres preguntas centrales: **¿en qué territorios se concentran los accidentes por MAP/MUSE?, ¿qué patrones espaciales pueden identificarse mediante estimación de densidad?, y ¿qué grupos de municipios presentan perfiles similares de afectación?** Con ello, se espera aportar una lectura técnica que contribuya a la priorización territorial, la prevención del riesgo y la toma de decisiones basada en evidencia en contextos humanitarios.

### **6. KDE / estimación de densidad espacial**


```python
### Librerías necesarias para KDE y visualización espacial
from sklearn.neighbors import KernelDensity
from sklearn.model_selection import GridSearchCV
from pyproj import Transformer
import geopandas as gpd
from shapely.geometry import Point
```

### Preparación de coordenadas

En este proyecto, se decidió extraer y estructurar exclusivamente las columnas de latitud y longitud de los accidentes por MAP/MUSE. Esta etapa es vital para garantizar la integridad de la estimación, ya que cualquier registro incompleto o con errores de georreferenciación podría introducir sesgos significativos en el cálculo de las densidades.

```python
### Creamos una copia de la base para el análisis espacial
df_kde = df.copy()

### Conservamos únicamente registros con coordenadas válidas
df_kde = df_kde.dropna(subset=["latitude", "longitude"])

### Filtramos coordenadas dentro de rangos geográficos posibles
df_kde = df_kde[
    (df_kde["latitude"].between(-5, 15)) &
    (df_kde["longitude"].between(-85, -65))
].copy()

### Verificamos la cantidad de eventos disponibles para el análisis espacial
print("Número de eventos con coordenadas válidas:", df_kde.shape[0])
```

### Transformación de coordenadas

Una de las decisiones técnicas que se toma en este apartado es la transformación de las coordenadas desde un sistema geográfico (WGS84 en grados) hacia un sistema de coordenadas proyectadas (métrico). Esta elección se fundamenta en que los grados decimales son unidades no lineales que varían su distancia física dependiendo de la latitud, lo cual distorsionaría el cálculo del Kernel. Al proyectar los datos a un plano métrico, se asegura que el análisis de proximidad y la medición de distancias entre accidentes sean constantes y precisos en todo el país.

```python
### Transformamos las coordenadas de latitud-longitud a metros
transformer = Transformer.from_crs("EPSG:4326", "EPSG:3857", always_xy=True)

### Convertimos longitud y latitud a coordenadas x, y en metros
x_metros, y_metros = transformer.transform(
    df_kde["longitude"].values,
    df_kde["latitude"].values
)

### Creamos una matriz de coordenadas para KDE
coords_metros = np.vstack([x_metros, y_metros]).T

### Verificamos la dimensión de la matriz de coordenadas
print("Dimensión de la matriz de coordenadas:", coords_metros.shape)
```

### Selección del ancho de banda
A continuación, se desea equilibrar el sesgo y la varianza, ya que un ancho de banda demasiado estrecho produciría una superficie ruidosa centrada solo en puntos individuales, mientras que uno muy amplio ocultaría patrones locales críticos. En ese sentido, se optó por un criterio de selección optimizado que permite capturar la estructura regional de los "puntos calientes" sin perder la especificidad de los focos de accidentalidad, garantizando que el modelo revele la tendencia de la amenaza de manera clara y estadísticamente significativa.

```python
### Definimos valores posibles de ancho de banda en metros
### 20000 equivale a 20 km, 50000 a 50 km, etc.
bandwidths = {
    "bandwidth": [20000, 30000, 50000, 75000, 100000, 150000]
}

### Aplicamos búsqueda por validación cruzada para seleccionar el ancho de banda
grid_search = GridSearchCV(
    KernelDensity(kernel="gaussian"),
    bandwidths,
    cv=5
)

### Ajustamos la búsqueda sobre las coordenadas proyectadas
grid_search.fit(coords_metros)

### Extraemos el mejor ancho de banda
mejor_bandwidth = grid_search.best_params_["bandwidth"]

print("Mejor ancho de banda seleccionado:", mejor_bandwidth, "metros")
print("Mejor ancho de banda seleccionado en Km:", mejor_bandwidth / 1000, "km")
```

### Ajuste del modelo KDE

La elección de la función gaussiana para el Kernel se justifica por la capacidad de modelar la influencia de cada accidente de manera suave y decreciente a medida que nos alejamos del epicentro de eventos. Durante el ajuste, el algoritmo distribuye la probabilidad sobre el espacio transformado, permitiendo que la superposición de múltiples eventos cercanos genere los picos de densidad que definen las zonas de mayor riesgo. Este proceso de ajuste es fundamental para la inferencia espacial, ya que permite estimar la intensidad de la amenaza incluso en áreas donde no se han registrado accidentes recientes pero que se encuentran dentro de la zona de influencia de focos conocidos.

```python
### Ajustamos el modelo KDE con el ancho de banda seleccionado
kde = KernelDensity(
    bandwidth=mejor_bandwidth,
    kernel="gaussian"
)

kde.fit(coords_metros)
```

### Crear grilla espacial para evaluar la densidad

```python
### Definimos un margen espacial alrededor de los puntos observados
margen = 50000

### Creamos una grilla regular dentro del rango observado de coordenadas
x_grid = np.linspace(
    coords_metros[:, 0].min() - margen,
    coords_metros[:, 0].max() + margen,
    300
)

y_grid = np.linspace(
    coords_metros[:, 1].min() - margen,
    coords_metros[:, 1].max() + margen,
    300
)

### Combinamos los valores de "x" y "y" para crear la grilla espacial
xx, yy = np.meshgrid(x_grid, y_grid)

### Convertimos la grilla en una matriz de puntos
grid_metros = np.vstack([
    xx.ravel(),
    yy.ravel()
]).T

### Evaluamos la densidad KDE en cada punto de la grilla
log_densidad = kde.score_samples(grid_metros)

### Convertimos la densidad logarítmica a escala normal
densidad = np.exp(log_densidad)

### Reorganizamos la densidad con la forma de la grilla
densidad_grid = densidad.reshape(xx.shape)
```

### Gráfico de densidad espacial KDE

```python
### Graficamos la superficie de densidad espacial
plt.figure(figsize=(10, 10))

### Mapa de calor de la densidad estimada
plt.contourf(
    xx,
    yy,
    densidad_grid,
    levels=30,
    alpha=0.75
)

### Agregamos curvas de nivel
plt.contour(
    xx,
    yy,
    densidad_grid,
    levels=10,
    linewidths=0.8
)

### Agregamos los puntos de eventos observados
plt.scatter(
    coords_metros[:, 0],
    coords_metros[:, 1],
    s=3,
    alpha=0.25
)

plt.xlabel("Coordenada X en metros")
plt.ylabel("Coordenada Y en metros")
plt.title("Estimación de densidad espacial KDE de eventos MAP/MUSE")
plt.tight_layout()
plt.show()
```

### Gráfico en coordenadas geográficas

Al situar las densidades calculadas dentro de los límites políticos y geográficos de Colombia, el modelo adquiere un contexto real que permite identificar no solo la magnitud del peligro, sino los activos estratégicos y las comunidades que se encuentran bajo amenaza. Esta visualización permite correlacionar los hotspots con elementos geográficos como cordilleras, ríos y vías principales, facilitando la interpretación de por qué ciertas zonas presentan una mayor acumulación de eventos.

```python
### Transformamos la grilla de metros nuevamente a longitud-latitud
transformer_inv = Transformer.from_crs("EPSG:3857", "EPSG:4326", always_xy=True)

lon_grid, lat_grid = transformer_inv.transform(xx, yy)

### Graficamos KDE en coordenadas geográficas
plt.figure(figsize=(10, 10))

plt.contourf(
    lon_grid,
    lat_grid,
    densidad_grid,
    levels=30,
    alpha=0.75
)

plt.contour(
    lon_grid,
    lat_grid,
    densidad_grid,
    levels=10,
    linewidths=0.8
)

plt.scatter(
    df_kde["longitude"],
    df_kde["latitude"],
    s=3,
    alpha=0.25
)

plt.xlabel("Longitud")
plt.ylabel("Latitud")
plt.title("Densidad espacial estimada de eventos MAP/MUSE en Colombia")
plt.tight_layout()
plt.show()
```

### KDE diferenciado por MAP y MUSE

A continuación, se pretende observar las diferencias de ambos incidentes. Al comparar visualmente ambos mapas de calor, se puede llegar a detectar si existe una superposición geográfica o si, por el contrario, cada amenaza domina sectores distintos.

```python
### Función para graficar KDE por tipo de evento
def graficar_kde_tipo_evento(df_base, tipo_evento, bandwidth=50000):

    ### Filtramos el tipo de evento
    df_tipo = df_base[df_base["Tipo de evento"] == tipo_evento].copy()

    ### Transformamos coordenadas a metros
    x_tipo, y_tipo = transformer.transform(
        df_tipo["longitude"].values,
        df_tipo["latitude"].values
    )

    coords_tipo = np.vstack([x_tipo, y_tipo]).T

    ### Ajustamos KDE
    kde_tipo = KernelDensity(
        bandwidth=bandwidth,
        kernel="gaussian"
    )

    kde_tipo.fit(coords_tipo)

    ### Evaluamos la densidad en la misma grilla
    log_densidad_tipo = kde_tipo.score_samples(grid_metros)
    densidad_tipo = np.exp(log_densidad_tipo).reshape(xx.shape)

    ### Graficamos
    plt.figure(figsize=(10, 10))

    plt.contourf(
        lon_grid,
        lat_grid,
        densidad_tipo,
        levels=30,
        alpha=0.75
    )

    plt.contour(
        lon_grid,
        lat_grid,
        densidad_tipo,
        levels=10,
        linewidths=0.8
    )

    plt.scatter(
        df_tipo["longitude"],
        df_tipo["latitude"],
        s=3,
        alpha=0.25
    )

    plt.xlabel("Longitud")
    plt.ylabel("Latitud")
    plt.title(f"Densidad espacial KDE - {tipo_evento}")
    plt.tight_layout()
    plt.show()
```

### KDE para accidentes por MAP

Debido a que las MAP representan la gran mayoría de los eventos en el dataset, este gráfico de densidad es el que presenta los focos más intensos y definidos del país. Se puede observar que los picos de densidad en este mapa señalan áreas donde la presencia de minas puede representar una barrera estructural para la movilidad y el uso productivo de la tierras.

```python
### KDE para accidentes por MAP
graficar_kde_tipo_evento(
    df_kde,
    "Accidente por MAP",
    bandwidth=mejor_bandwidth
)
```

### **4. Clustering municipal con PCA + K-means**

### Construcción de la base municipal

La construcción de la base municipal consiste en un proceso de agregación de datos crudos en un perfil de riesgo multidimensional por territorio. En este sentido, se definieron variables como 'Intensidad Reciente' y 'Persistencia'. La intensidad reciente captura la dinámica del riesgo en el corto plazo, permitiendo identificar focos emergentes de accidentes, mientras que la persistencia mide la frecuencia histórica y la recurrencia del fenómeno. Esta distinción es fundamental ya que permite al modelo diferenciar entre municipios con una crisis coyuntural y aquellos donde la presencia de MAP/MUSE es un problema estructural y crónico, dotando una profundidad temporal necesaria para un agrupamiento.

```python
### Librerías necesarias
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
### Creamos una copia de la base para el clustering
df_cluster = df.copy()

### Creamos variables binarias para facilitar la agregación municipal
df_cluster["es_map"] = np.where(df_cluster["Tipo de evento"] == "Accidente por MAP", 1, 0)
df_cluster["es_muse"] = np.where(df_cluster["Tipo de evento"] == "Accidente por MUSE", 1, 0)
df_cluster["es_rural"] = np.where(df_cluster["Tipo de area"] == "Rural", 1, 0)

### Construimos una base agregada a nivel departamento-municipio
base_municipal = (
    df_cluster
    .groupby(["departamento", "municipio"])
    .agg(
        total_eventos=("Tipo de evento", "size"),
        eventos_map=("es_map", "sum"),
        eventos_muse=("es_muse", "sum"),
        eventos_rurales=("es_rural", "sum"),
        anio_min=("Año", "min"),
        anio_max=("Año", "max"),
        anios_con_evento=("Año", "nunique"),
        latitud_promedio=("latitude", "mean"),
        longitud_promedio=("longitude", "mean")
    )
    .reset_index()
)

base_municipal.head()
```

### Variables de intensidad reciente y persistencia

La intensidad reciente permite capturar la dinámica actual del conflicto y el riesgo emergente, identificando dónde se han concentrado los accidentes en el periodo más cercano (2020 al presente año). Por otro lado, la persistencia mide la frecuencia histórica y la recurrencia del peligro a lo largo del tiempo. 

```python
### Calculamos eventos recientes desde 2020
eventos_recientes = (
    df_cluster[df_cluster["Año"] >= 2020]
    .groupby(["departamento", "municipio"])
    .size()
    .reset_index(name="eventos_recientes")
)

### Unimos la información reciente a la base municipal
base_municipal = base_municipal.merge(
    eventos_recientes,
    on=["departamento", "municipio"],
    how="left"
)

### Reemplazamos municipios sin eventos recientes por cero
base_municipal["eventos_recientes"] = base_municipal["eventos_recientes"].fillna(0)
```

```python
### Calculamos el máximo anual de eventos por municipio
eventos_anuales_municipio = (
    df_cluster
    .groupby(["departamento", "municipio", "Año"])
    .size()
    .reset_index(name="eventos_anio")
)

max_eventos_anio = (
    eventos_anuales_municipio
    .groupby(["departamento", "municipio"])["eventos_anio"]
    .max()
    .reset_index(name="max_eventos_anio")
)

### Unimos el máximo anual a la base municipal
base_municipal = base_municipal.merge(
    max_eventos_anio,
    on=["departamento", "municipio"],
    how="left"
)
```

```python
### Creamos indicadores relativos
base_municipal["pct_map"] = base_municipal["eventos_map"] / base_municipal["total_eventos"]
base_municipal["pct_muse"] = base_municipal["eventos_muse"] / base_municipal["total_eventos"]
base_municipal["pct_rural"] = base_municipal["eventos_rurales"] / base_municipal["total_eventos"]
base_municipal["pct_reciente"] = base_municipal["eventos_recientes"] / base_municipal["total_eventos"]

### Promedio anual de eventos en los años donde el municipio registra ocurrencia
base_municipal["promedio_anual_eventos"] = (
    base_municipal["total_eventos"] / base_municipal["anios_con_evento"]
)

### Transformaciones logarítmicas para reducir el efecto de valores extremos
base_municipal["log_total_eventos"] = np.log1p(base_municipal["total_eventos"])
base_municipal["log_eventos_recientes"] = np.log1p(base_municipal["eventos_recientes"])
base_municipal["log_max_eventos_anio"] = np.log1p(base_municipal["max_eventos_anio"])
base_municipal["log_promedio_anual"] = np.log1p(base_municipal["promedio_anual_eventos"])

base_municipal.head()
```

### Variables para el clustering

La selección de las variables finales para alimentar el algoritmo de clustering se basó en un criterio de representatividad y relevancia multidimensional. La selección de variables es lo que permite que el análisis de componentes principales (PCA) posterior pueda extraer dimensiones latentes con un alto valor explicativo.

```python
### Seleccionamos las variables que resumen el perfil de afectación municipal
variables_cluster = [
    "log_total_eventos",
    "log_eventos_recientes",
    "pct_reciente",
    "pct_rural",
    "anios_con_evento",
    "log_max_eventos_anio",
    "log_promedio_anual",
    "pct_muse"
]

### Creamos la matriz de análisis
X_cluster = base_municipal[variables_cluster].copy()

### Reemplazamos posibles valores infinitos o faltantes
X_cluster = X_cluster.replace([np.inf, -np.inf], np.nan).fillna(0)

### Estandarizamos las variables (Esto es necesario porque las variables están en escalas diferentes)
scaler = StandardScaler()
X_cluster_scaled = scaler.fit_transform(X_cluster)
```

### PCA para reducir dimensionalidad

Dado que el perfil municipal se compone de múltiples variables que presentan altos grados de correlación entre sí (como la cantidad de víctimas, el tipo de área y la frecuencia anual), se implementó el Análisis de Componentes Principales (PCA). Al transformar las variables originales en un conjunto de componentes ortogonales, logramos retener la mayor proporción de la varianza explicada del fenómeno con un menor número de dimensiones. Este paso no solo optimiza el rendimiento computacional del modelo de clúster, sino que permite que el agrupamiento se realice sobre las 'estructuras latentes' del riesgo, asegurando que la separación entre municipios se base en patrones de fondo y no en el ruido estadístico de variables altamente correlacionadas.


```python
### Aplicamos PCA para revisar la varianza explicada por cada componente
pca_full = PCA()
pca_full.fit(X_cluster_scaled)

### Calculamos varianza explicada y acumulada
varianza_pca = pd.DataFrame({
    "componente": np.arange(1, len(pca_full.explained_variance_ratio_) + 1),
    "varianza_explicada": pca_full.explained_variance_ratio_,
    "varianza_acumulada": np.cumsum(pca_full.explained_variance_ratio_)
})

varianza_pca
```

```python
### Graficamos la varianza explicada acumulada
plt.figure(figsize=(8, 5))

plt.plot(
    varianza_pca["componente"],
    varianza_pca["varianza_acumulada"],
    marker="o"
)

plt.axhline(y=0.85, linestyle="--", label="85% de varianza")
plt.xlabel("Número de componentes principales")
plt.ylabel("Varianza explicada acumulada")
plt.title("Varianza explicada acumulada - PCA municipal")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

Teniendo en cuenta la cantidad de componentes que capturan aproximadamente el 85%, se seleccionó la cantidad de 3. Asi mismo se puede observar de manera detallada por la regla del "codo".

```python
### Seleccionamos el número mínimo de componentes que explican al menos el 80% de la varianza
umbral_pca = 0.80

k_pca = (
    np.argmax(varianza_pca["varianza_acumulada"].values >= umbral_pca) + 1
)

print("Número de componentes seleccionados:", k_pca)
print("Varianza acumulada explicada:", round(varianza_pca.loc[k_pca - 1, "varianza_acumulada"] * 100, 2), "%")
```

```python
### Ajustamos PCA con el número de componentes seleccionados
pca = PCA(n_components=k_pca)

X_pca = pca.fit_transform(X_cluster_scaled)

### Convertimos los componentes principales en DataFrame
componentes_pca = pd.DataFrame(
    X_pca,
    columns=[f"PC{i+1}" for i in range(k_pca)]
)

componentes_pca.head()
```

```python
### Revisamos los pesos de cada variable en los componentes principales
loadings = pd.DataFrame(
    pca.components_.T,
    index=variables_cluster,
    columns=[f"PC{i+1}" for i in range(k_pca)]
)

loadings
```

### Selección del número de clústeres con silueta

La determinación del número óptimo de clústeres no se realizó de manera arbitraria, sino que se fundamentó en el análisis del Coeficiente de Silueta, una métrica de validación interna que evalúa la consistencia de la agrupación

```python
### Evaluamos diferentes números de clústeres usando el coeficiente de silueta
resultados_k = []

for k in range(2, 9):

    modelo_kmeans = KMeans(
        n_clusters=k,
        random_state=123,
        n_init=25
    )

    etiquetas = modelo_kmeans.fit_predict(X_pca)

    silueta = silhouette_score(X_pca, etiquetas)

    resultados_k.append({
        "k": k,
        "silueta": silueta
    })

resultados_k = pd.DataFrame(resultados_k)

resultados_k
```

```python
### Graficamos el coeficiente de silueta según el número de clústeres
plt.figure(figsize=(8, 5))

plt.plot(
    resultados_k["k"],
    resultados_k["silueta"],
    marker="o"
)

plt.xlabel("Número de clústeres")
plt.ylabel("Coeficiente de silueta")
plt.title("Selección del número de clústeres")
plt.grid(True)
plt.tight_layout()
plt.show()
```

Al observar la curva generada por la gráfica anterior, se identifica el número de clústeres que maximiza la estructura de los datos. El punto máximo de esta gráfica representa la configuración donde los municipios están mejor asignados a sus respectivos perfiles de riesgo, minimizando la ambigüedad en la clasificación. Seleccionar el $K$ con el valor de 4 permite que el modelo de K-means capture de manera efectiva las disparidades territoriales entre zonas de alta persistencia histórica y focos de intensidad reciente.

```python
### Seleccionamos el número de clústeres con mayor coeficiente de silueta
k_optimo = 4
```

### Ajuste del modelo K-means

```python
### Ajustamos el modelo final de K-means
kmeans_final = KMeans(
    n_clusters=k_optimo,
    random_state=123,
    n_init=25
)

### Asignamos cada municipio a un clúster
base_municipal["cluster"] = kmeans_final.fit_predict(X_pca)

### Agregamos los dos primeros componentes para visualización
base_municipal["PC1"] = X_pca[:, 0]
base_municipal["PC2"] = X_pca[:, 1]

base_municipal.head()
```

### **5. DBSCAN sobre puntos de eventos**


La implementación de DBSCAN sobre las coordenadas de los eventos de MAP/MUSE responde a la necesidad de identificar focos de riesgo que no se ajustan a formas geométricas regulares ni a límites administrativos. Esta técnica de aprendizaje no supervisado es particularmente robusta frente a los valores atípicos (outliers), permitiendo que el análisis se concentre exclusivamente en las zonas donde la acumulación de eventos sugiere una actividad minada sistemática, separándola de los incidentes aislados que podrían distorsionar la interpretación de los puntos calientes.

```python
from sklearn.cluster import DBSCAN
from pyproj import Transformer
```

```python
### Creamos una copia de eventos con coordenadas válidas
df_dbscan = df.dropna(subset=["latitude", "longitude"]).copy()

### Transformamos coordenadas a metros
transformer = Transformer.from_crs("EPSG:4326", "EPSG:3857", always_xy=True)

x_db, y_db = transformer.transform(
    df_dbscan["longitude"].values,
    df_dbscan["latitude"].values
)

coords_dbscan = np.vstack([x_db, y_db]).T
```

```python
### Aplicamos DBSCAN
### eps=50000 equivale a 50 km
### min_samples=20 exige al menos 20 eventos cercanos para formar un núcleo
dbscan = DBSCAN(
    eps=50000,
    min_samples=20
)

df_dbscan["cluster_dbscan"] = dbscan.fit_predict(coords_dbscan)

### Número de clústeres encontrados, excluyendo ruido
n_clusters_dbscan = len(set(df_dbscan["cluster_dbscan"])) - (1 if -1 in df_dbscan["cluster_dbscan"].values else 0)

print("Número de clústeres espaciales encontrados:", n_clusters_dbscan)
print("Número de eventos clasificados como ruido:", (df_dbscan["cluster_dbscan"] == -1).sum())
```

```python
### Convertimos a GeoDataFrame
eventos_dbscan_gdf = gpd.GeoDataFrame(
    df_dbscan,
    geometry=gpd.points_from_xy(
        df_dbscan["longitude"],
        df_dbscan["latitude"]
    ),
    crs="EPSG:4326"
)

### Graficamos los clústeres espaciales DBSCAN
fig, ax = plt.subplots(figsize=(10, 12))

colombia.plot(
    ax=ax,
    color="whitesmoke",
    edgecolor="black",
    linewidth=0.8
)

eventos_dbscan_gdf.plot(
    ax=ax,
    column="cluster_dbscan",
    categorical=True,
    legend=True,
    cmap="Dark2",
    markersize=10,
    alpha=0.6
)

xmin, ymin, xmax, ymax = colombia.total_bounds
ax.set_xlim(xmin, xmax)
ax.set_ylim(ymin, ymax)

ax.set_title("Clústeres espaciales de eventos MAP/MUSE con DBSCAN")
ax.set_xlabel("Longitud")
ax.set_ylabel("Latitud")

plt.tight_layout()
plt.show()
```

### **6. APRENDIZAJE SUPERVISADO (RANDOM FOREST)**

Al entrenar un modelo de ensamble basado en múltiples árboles de decisión, se busca construir un motor predictivo capaz de asignar automáticamente un perfil de riesgo a cualquier unidad territorial con base en sus características observables. Esta transición es fundamental para la escalabilidad del proyecto, ya que se dieferenciaría al ser una herramienta capaz de clasificar municipios nuevos o con información incompleta en la base de datos, garantizando que la priorización territorial no dependa únicamente de un historial estático, sino de un modelo dinámico de inferencia.


```python
### APRENDIZAJE SUPERVISADO: CLASIFICACIÓN CON RANDOM FOREST Y SHAP
import seaborn as sns
from sklearn.model_selection import train_test_split, RandomizedSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
import shap ### Explicabilidad

### Seleccionamos las variables predictoras (X) y la variable objetivo (y)
variables_predictoras = [
    'total_eventos',
    'pct_rural',
    'anios_con_evento',
    'eventos_recientes',
    'pct_reciente'
]

X = base_municipal[variables_predictoras]
y = base_municipal['cluster'] # La etiqueta generada por K-means

### División estratificada para mantener la proporción de clústeres en train y test
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

### OPTIMIZACIÓN DE HIPERPARÁMETROS (RandomizedSearchCV)
### Definimos la grilla de búsqueda. Usamos class_weight='balanced' para
### penalizar los errores en las clases minoritarias (ej. Clúster de Alta Prioridad)
param_grid = {
    'n_estimators': [100, 200, 300, 500],
    'max_depth': [None, 5, 10, 15],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'class_weight': ['balanced', 'balanced_subsample']
}

### Inicializamos el modelo base
rf_base = RandomForestClassifier(random_state=42)

### Configuramos la búsqueda aleatoria con validación cruzada (CV=5)
rf_random = RandomizedSearchCV(
    estimator=rf_base,
    param_distributions=param_grid,
    n_iter=20, # Número de combinaciones a probar
    cv=5,      # 5 Folds para Cross-Validation
    verbose=1,
    random_state=42,
    n_jobs=-1  # Usa todos los núcleos disponibles del procesador
)

### Entrenamos buscando los mejores hiperparámetros
print("Buscando los mejores hiperparámetros...")
rf_random.fit(X_train, y_train)

### Obtenemos el mejor modelo
best_rf = rf_random.best_estimator_
print(f"\nMejores parámetros encontrados: {rf_random.best_params_}")

### EVALUACIÓN DEL MODELO
### Predicciones sobre el conjunto de prueba
y_pred = best_rf.predict(X_test)

print("\n=== REPORTE DE CLASIFICACIÓN ===")
print(classification_report(y_test, y_pred))

### Matriz de Confusión
plt.figure(figsize=(8, 6))
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=np.unique(y), yticklabels=np.unique(y))
plt.title('Matriz de Confusión - Random Forest')
plt.ylabel('Clúster Real')
plt.xlabel('Clúster Predicho')
plt.tight_layout()
plt.show()

###INTERPRETABILIDAD AVANZADA CON SHAP
print("\nGenerando explicabilidad con SHAP")

explainer = shap.TreeExplainer(best_rf)
shap_values = explainer.shap_values(X_test)

### Gráfico 1: Importancia global de las variables
plt.title("Importancia Global de Variables (SHAP)")
shap.summary_plot(shap_values, X_test, plot_type="bar", show=True)

### Gráfico 2: Impacto detallado (Beeswarm plot)
clase_a_explicar = 0
print(f"\nImpacto direccional de las variables para predecir el Clúster {np.unique(y)[clase_a_explicar]}")

### Ajuste dinámico de dimensiones según la versión de SHAP o cantidad de clústeres
import numpy as np

if isinstance(shap_values, list):
    ### Versiones antiguas de SHAP (Multiclase devuelve lista)
    valores_graficar = shap_values[clase_a_explicar]
elif len(np.shape(shap_values)) == 3:
    ### Versiones nuevas de SHAP (Multiclase devuelve arreglo 3D)
    valores_graficar = shap_values[:, :, clase_a_explicar]
else:
    ### Si es binario (solo 2 clústeres), devuelve arreglo 2D directo
    valores_graficar = shap_values

### Generar el gráfico con los valores corregidos
shap.summary_plot(valores_graficar, X_test, show=True)
```

La evaluación del clasificador Random Forest revela una alta consistencia entre las características observadas de los municipios y su asignación a los perfiles de riesgo. Las métricas de precisión, sensibilidad (recall) y F1-score alcanzadas demuestran que el modelo posee una capacidad predictiva robusta para distinguir entre los diferentes niveles de afectación. Es particularmente relevante el alto valor de recall en los clústeres de alta prioridad, lo que indica que el algoritmo minimiza los falsos negativos; es decir, asegura que los municipios con una problemática crítica de MAP/MUSE no sean clasificados erróneamente como zonas de bajo riesgo.

En cambio, al evaluar el reporte de clasificación, se evidencia un fenómeno estadístico particular en el Clúster 3, el cual presenta una precisión baja (0.29) contrastada con una sensibilidad o recall perfecto (1.00). Este comportamiento es un resultado directo de la técnica de balanceo de clases (class_weight='balanced') aplicada para compensar la minoría muestral de este grupo (11). El modelo ha desarrollado un sesgo conservador ya que al intentar minimizar a cero los falsos negativos, el algoritmo sacrifica su precisión generando una alta tasa de falsos positivos.

La matriz de confusión permite visualizar la precisión del modelo en la identificación de cada perfil de riesgo. La fuerte concentración de valores en la diagonal principal corrobora la estabilidad del algoritmo y su capacidad de generalización. Los escasos errores de clasificación observados ocurren generalmente entre clústeres con características colindantes (Clúster 1 vs Clúster 3), lo cual es esperado dada la similitud en naturaleza continua del fenómeno del conflicto y la baja clasifición de los municiíos por la variable de ruralidad. Sin embargo, la capacidad del modelo para separar claramente los extremos (municipios con afectación esporádica frente a aquellos con crisis humanitarias persistentes) garantiza que la herramienta sea confiable para la toma de decisiones presupuestales y operativas en zonas confictivas y con mayor indice de eventos MAP/MUSE. Esta matriz actúa como una prueba de validación cruzada que confirma la coherencia de la lógica de agrupamiento aplicada al territorio nacional.
